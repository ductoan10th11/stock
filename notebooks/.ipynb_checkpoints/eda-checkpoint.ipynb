# EDA dữ liệu nến ngày (1D) – TCB, CTG, VNM, ACB

Notebook khảo sát **dữ liệu thô** (`data/raw/1D/*.csv`, thu thập bằng `crawl.py`). Phần tương quan đặc trưng – nhãn sẽ bổ sung sau khi có bước tiền xử lý.

Chạy: từ thư mục gốc dự án, `.venv/bin/jupyter notebook notebooks/eda.ipynb`.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

SEED = 42
SYMBOLS = ["TCB", "CTG", "VNM", "ACB"]
THETA = 0.005  # ngưỡng nhãn 0.5% (docs/data-preprocess.md)

ROOT = Path.cwd()
if not (ROOT / "data").exists():
    ROOT = ROOT.parent
RAW = ROOT / "data" / "raw" / "1D"

plt.rcParams.update({"figure.dpi": 110, "axes.grid": True, "grid.alpha": 0.3})
raw = {s: pd.read_csv(RAW / f"{s}.csv", parse_dates=["time"]) for s in SYMBOLS}
print("Thư mục dữ liệu:", RAW)
```

## 1. Tổng quan dữ liệu

```python
rows = []
for s, d in raw.items():
    rows.append({"Mã": s, "Số nến": len(d), "Từ": d["time"].min().date(),
                 "Đến": d["time"].max().date(), "Số cột": d.shape[1]})
pd.DataFrame(rows).set_index("Mã")
```

```python
raw['TCB'].dtypes.to_frame('Kiểu dữ liệu').T
```

## 2. Thống kê mô tả

```python
stats = pd.concat({s: d.drop(columns="time").describe().T for s, d in raw.items()})
stats.round(2)
```

## 3. Chất lượng dữ liệu

```python
rows = []
for s, d in raw.items():
    bad = ((d.open > d.high) | (d.close > d.high) | (d.open < d.low) | (d.close < d.low))
    rows.append({
        "Mã": s,
        "Thiếu (ô)": int(d.isna().sum().sum()),
        "Ngày trùng": int(d.time.duplicated().sum()),
        "OHLC sai logic": int(bad.sum()),
        "Volume = 0": int((d.volume == 0).sum()),
        "Sắp xếp tăng dần": bool(d.time.is_monotonic_increasing),
    })
quality = pd.DataFrame(rows).set_index("Mã")
quality
```

```python
# Chi tiết các ngày có vấn đề
for s, d in raw.items():
    dup = d[d.time.duplicated(keep=False)]
    bad = d[((d.open > d.high) | (d.close > d.high) | (d.open < d.low) | (d.close < d.low))]
    if len(dup): print(s, "ngày trùng:", sorted(dup.time.dt.date.unique()))
    if len(bad): print(s, "OHLC sai:", [str(x.date()) for x in bad.time])
```

## 4. Giá đóng cửa và khối lượng theo thời gian

```python
fig, axes = plt.subplots(4, 2, figsize=(13, 11), sharex="col")
for i, (s, d) in enumerate(raw.items()):
    axes[i, 0].plot(d.time, d.close, lw=1)
    axes[i, 0].set_ylabel(f"{s} – giá (nghìn đồng)")
    axes[i, 1].bar(d.time, d.volume / 1e6, width=1.5)
    axes[i, 1].set_ylabel("Volume (triệu cp)")
axes[0, 0].set_title("Giá đóng cửa"); axes[0, 1].set_title("Khối lượng")
plt.tight_layout(); plt.show()
```

## 5. Phân bố lợi suất ngày

```python
ret = {}
for s, d in raw.items():
    g = d.groupby("time", as_index=False).agg(close=("close", "last"))
    r = np.log(g.close / g.close.shift(1)).dropna()
    ret[s] = pd.Series(r.values, index=g.time[1:])
summary = pd.DataFrame({
    "Trung bình (%)": {s: r.mean() * 100 for s, r in ret.items()},
    "Độ lệch chuẩn (%)": {s: r.std() * 100 for s, r in ret.items()},
    "Độ lệch (skew)": {s: r.skew() for s, r in ret.items()},
    "Độ nhọn dư (kurt)": {s: r.kurt() for s, r in ret.items()},
    "Min (%)": {s: r.min() * 100 for s, r in ret.items()},
    "Max (%)": {s: r.max() * 100 for s, r in ret.items()},
})
summary.round(3)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, (s, r) in zip(axes.ravel(), ret.items()):
    ax.hist(r * 100, bins=80, color="tab:blue", alpha=0.8)
    ax.axvline(THETA * 100, color="r", ls="--", lw=1); ax.axvline(-THETA * 100, color="r", ls="--", lw=1)
    ax.set_title(s); ax.set_xlabel("Lợi suất log ngày (%)"); ax.set_ylabel("Số phiên")
plt.tight_layout(); plt.show()
```

## 6. Phân bố nhãn theo ngưỡng θ

```python
def class_share(r, theta):
    nxt = r.shift(-1).dropna()  # lợi suất của phiên kế tiếp
    return pd.Series({"Giảm": (nxt < -theta).mean(), "Ngang": (nxt.abs() <= theta).mean(),
                      "Tăng": (nxt > theta).mean()})
share = pd.DataFrame({s: class_share(r, THETA) for s, r in ret.items()}).T * 100
share.round(1)
```

```python
thetas = [0.002, 0.003, 0.005, 0.01]
flat = pd.DataFrame({f"θ={t*100:.1f}%": {s: class_share(r, t)["Ngang"] * 100 for s, r in ret.items()}
                     for t in thetas})
print("Tỷ lệ lớp 'ngang' (%) theo ngưỡng θ:")
flat.round(1)
```

```python
ax = share.plot(kind="bar", stacked=True, figsize=(7, 4), rot=0, color=["tab:red", "tab:gray", "tab:green"])
ax.set_ylabel("Tỷ lệ (%)"); ax.set_xlabel("Mã"); ax.set_title(f"Phân bố nhãn, θ = {THETA*100:.1f}%")
plt.show()
```

## 7. Biến động theo thời gian và tự tương quan

```python
fig, ax = plt.subplots(figsize=(11, 4))
for s, r in ret.items():
    (r.rolling(20).std() * 100).plot(ax=ax, label=s, lw=1)
ax.set_ylabel("Độ lệch chuẩn 20 phiên (%)"); ax.legend(); plt.show()
```

```python
lags = range(1, 11)
ac = pd.DataFrame({s: [r.autocorr(l) for l in lags] for s, r in ret.items()}, index=lags)
ac_abs = pd.DataFrame({s: [r.abs().autocorr(l) for l in lags] for s, r in ret.items()}, index=lags)
fig, axes = plt.subplots(1, 2, figsize=(11, 3.8), sharey=True)
ac.plot(kind="bar", ax=axes[0], title="Tự tương quan của lợi suất"); axes[0].set_xlabel("Độ trễ (phiên)")
ac_abs.plot(kind="bar", ax=axes[1], title="Tự tương quan của |lợi suất|"); axes[1].set_xlabel("Độ trễ (phiên)")
plt.tight_layout(); plt.show()
```
